```python
import pandas as pd

df = pd.read_csv('/content/Unemployment in India.csv')
df.head()
```

```python
print(df.shape)
print(df.info())
print(df.isnull().sum())
```

```python
df.columns = df.columns.str.strip()
df = df.dropna()

print(df.isnull().sum())
print(df.shape)
```

```python
import matplotlib.pyplot as plt

df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)

df.groupby('Date')['Estimated Unemployment Rate (%)'].mean().plot()

plt.title('Unemployment Rate in India')
plt.xlabel('Date')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.show()
```

```python
highest = df.loc[df['Estimated Unemployment Rate (%)'].idxmax()]

print("State with highest unemployment rate:", highest['Region'])
print("Unemployment Rate:", highest['Estimated Unemployment Rate (%)'], "%")
print("Date:", highest['Date'])
```

```python
state_rate = df.groupby('Region')['Estimated Unemployment Rate (%)'].mean()

state_rate = state_rate.sort_values(ascending=False)

print(state_rate.head(10))

state_rate.head(10).plot(kind='bar')
plt.title('Top 10 Regions by Unemployment Rate')
plt.ylabel('Average Unemployment Rate (%)')
plt.show()
```

## Conclusion

The Unemployment Analysis project studies unemployment rates across different regions of India. The dataset was cleaned by removing missing values. Graphs were used to understand unemployment trends and compare regions. This analysis helps identify regions with higher unemployment rates.
